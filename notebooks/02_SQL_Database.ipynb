```python
import sqlite3
import pandas as pd
conn=sqlite3.connect("people_analytics.db")
clean_df=pd.read_sql("SELECT * FROM employees",conn)
```

```python
clean_df["Employee_ID"]=[f"{str(i).zfill(4)}"
  for i in range(1,len(clean_df)+1)]
```

```python
clean_df[["EmployeeNumber","Employee_ID"]].head()
```

```python
clean_df["Career_Stagnation_Index"]=(clean_df["YearsSinceLastPromotion"]/(clean_df["YearsAtCompany"]+1)).round(2)
```

```python
clean_df["Promotion_Delay_Risk"]=(clean_df["YearsSinceLastPromotion"]>=5).astype(int)
```

```python
clean_df["Workload_Risk"]=(
    (
        clean_df["OverTime_Flag"]==1
    )&
    (
        clean_df["WorkLifeBalance"]<=2
    )
).astype(int)
```

```python
clean_df["Compensation_Risk"]=(
    clean_df["Income_Band"]=="Low"
).astype(int)
```

```python
clean_df["Travel_Risk"]=(
    clean_df["BusinessTravel"]=="Travel_Frequently"
).astype(int)
```

```python
clean_df["Satisfaction_Risk"]=(
    (
        clean_df["JobSatisfaction"]<=2
    )&
    (
        clean_df["EnvironmentSatisfaction"]<=2
    )
).astype(int)
```

```python
risk_columns=[
    "Promotion_Delay_Risk",
    "Workload_Risk",
    "Compensation_Risk",
    "Travel_Risk",
    "Satisfaction_Risk"
]
clean_df["HR_Risk_Score"]=clean_df[risk_columns].sum(axis=1)
```

```python
clean_df["HR_Risk_Level"]=pd.cut(
    clean_df["HR_Risk_Score"],
    bins=[-0.1,1.5,3.5,5.5],
    labels=["Low","Medium","High"]
)
```

```python
clean_df[[
    "Employee_ID",
    "Career_Stagnation_Index",
    "Promotion_Delay_Risk",
    "Workload_Risk",
    "Compensation_Risk",
    "Travel_Risk",
    "Satisfaction_Risk",
    "HR_Risk_Score",
    "HR_Risk_Level"
]].head(10)
```

```python
risk_columns=[
    "Promotion_Delay_Risk",
    "Workload_Risk",
    "Compensation_Risk",
    "Travel_Risk",
    "Satisfaction_Risk"
]
clean_df[risk_columns].head(10)
```

```python
clean_df[risk_columns].isnull().sum()
```

```python
clean_df[risk_columns].dtypes
```

```python
clean_df["HR_Risk_Score"]=clean_df[risk_columns].fillna(0).sum(axis=1)
```

```python
clean_df[["HR_Risk_Score"]].head(10)
```

```python
clean_df["HR_Risk_Level"]=pd.cut(
    clean_df["HR_Risk_Score"],
    bins=[-0.1,1.5,3.5,5.5],
    labels=["Low","Medium","High"],
    include_lowest=True
)
```

```python
clean_df[["HR_Risk_Score","HR_Risk_Level"]].head(10)
```

```python
risk_df=clean_df[["Employee_ID",
                  "Career_Stagnation_Index",
                  "Promotion_Delay_Risk",
                  "Workload_Risk",
                  "Compensation_Risk",
                  "Travel_Risk",
                  "Satisfaction_Risk",
                  "HR_Risk_Score",
                  "HR_Risk_Level"]]
risk_df.to_sql(
    "employee_risk",
    conn,
    if_exists="replace",
    index=False
)
```

```python
pd.read_sql("SELECT * FROM employee_risk LIMIT 5;",conn)
```

```python
conn.execute("""
CREATE VIEW IF NOT EXISTS high_risk_employees AS
SELECT Employee_ID,HR_Risk_Score,HR_Risk_Level
FROM employee_risk
WHERE HR_Risk_Level='High';""")
```

```python
pd.read_sql("SELECT * FROM high_risk_employees;",conn)
```

```python
conn.execute("""
CREATE VIEW IF NOT EXISTS department_summary AS
SELECT Department, COUNT(*) AS Employee_Count, ROUND(AVG(MonthlyIncome),2) AS Average_Income, ROUND(AVG(JobSatisfaction),2) AS Average_Job_Satisfaction
FROM employees
GROUP BY Department;
""")
```

```python
pd.read_sql("SELECT * FROM department_summary;",conn)
```
