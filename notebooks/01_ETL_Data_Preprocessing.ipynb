```python
clean_df=df.copy()
```

```python
clean_df.columns
```

```python
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.replace(" ","_")
)
```

```python
text_columns = clean_df.select_dtypes(include='object').columns
for col in text_columns :
  clean_df[col]=clean_df[col].str.strip().str.title()
```

```python
clean_df.dtypes
```

```python
clean_df=clean_df.drop_duplicates()
```

```python
print(clean_df.shape)
```

```python
print(clean_df["EmployeeNumber"].is_unique)
```

```python
print("Rows : ",clean_df.shape[0])
print("Columns : ",clean_df.shape[1])
print("Missing Numbers : ",clean_df.isnull().sum())
print("Duplicated Values : ",clean_df.duplicated().sum())
print("Unique Employees : ",clean_df["EmployeeNumber"].is_unique)
```

```python
for col in clean_df.columns:
  if clean_df[col].nunique()==1:
    print(col)
```

```python
constant_columns = ["EmployeeCount","Over18","StandardHours"]
clean_df.drop(columns=constant_columns,inplace=True)
```

```python
clean_df["Attrition"].value_counts()
```

```python
clean_df["Attrition"].value_counts(normalize=True)*100
```

```python
categorical_columns = clean_df.select_dtypes(include="object").columns
for col in categorical_columns:
  print("="*50)
  print(col)
  print(clean_df[col].value_counts())
```

```python
clean_df.describe()
```

```python
quality_report={
    "Rows" : clean_df.shape[0],
    "Columns" : clean_df.shape[1],
    "Missing Values" : int(clean_df.isnull().sum().sum()),
    "Duplicate Values" : int(clean_df.duplicated().sum()),
    "Constant Columns Removed" : len(constant_columns)
}
quality_report
```

```python
clean_df["Age_Group"]=pd.cut(
    clean_df["Age"],
    bins=[17,25,35,45,55,100],
    labels=["18-25","26-35","36-45","46-55","56+"]
)
```

```python
clean_df[["Age","Age_Group"]].head(10)
```

```python
clean_df["Age_Group"].value_counts()
```

```python
clean_df["Income_Band"]=pd.qcut(
    clean_df["MonthlyIncome"],
    q=4,
    labels=["Low","Medium","High","Executive"]
)
```

```python
clean_df["Income_Band"].value_counts()
```

```python
clean_df["Tenure_Bucket"]=pd.cut(
    clean_df["YearsAtCompany"],
    bins=[-1,2,5,10,40],
    labels=["0-2 Years","3-5 Years","6-10 Years","10+ Years"]
)
```

```python
clean_df[["YearsAtCompany","Tenure_Bucket"]].head(10)
```

```python
clean_df["Attrition_Flag"]=clean_df["Attrition"].map({"Yes":1,"No":0})
```

```python
clean_df["OverTime_Flag"]=clean_df["OverTime"].map({"Yes":1,"No":0})
```

```python
clean_df.info()
```

```python
clean_df.drop(columns=["Salary_Band"],inplace=True)
```

```python
clean_df.info()
```

```python
clean_df.to_csv("hr_cleaned.csv",index=False)
```
